Two sum array with brute force O(n^2) time and O(1) space complexity

```python
def twosum(arr, target):
  n = len(arr)
  for i in range(n):
    for j in range(i+1, n):
      if target == arr[i] + arr[j]:
        return [i,j]

  return []
```

```python
res=twosum([2,2,1,3,4],3)
```

```python
print(res)
```

Two sum array with hashmap O(n) time and O(n) space complexity

```python
def twoSumMemo(arr, target):
  hashmap ={}                 #map which stores arr element->key and its index ->value

  for i, num in enumerate(arr):
    complement = target - num
    if complement in hashmap:
      return [hashmap[complement], i]
    hashmap[num]=i
  return []


```

```python
print(twoSumMemo([2,2,1,3,4],3))
```

Two sum array with left and right pointers/ Two pointers O(n) time and O(1) space complexity

```python
def twoSumWithPointers(arr, target):
  left = 0
  right = len(arr)-1

  while left < right:
    if arr[left] + arr[right] == target:
      return [left, right]
    elif arr[left] + arr[right] < target:
      left += 1
    else:
      right -=1
  return []

```

```python
print(twoSumWithPointers([2,2,1,3,4],3))
```
